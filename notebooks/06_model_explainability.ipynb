# Notebook 06: Model Explainability & SHAP Analysis

## Business Context & Model Transparency
Financial credit underwriting requires transparent and explainable risk scoring. Regulators and credit committees demand clear explanations of *why* a particular borrower was classified as high risk or declined. We utilize **SHAP (SHapley Additive exPlanations)** to break down predictions into marginal feature contributions.

```python
import sys
import os
import json
import pandas as pd

sys.path.append(os.path.abspath('..'))
from ml.explain import get_model_explainer
from ml.predict import get_default_predictor

explainer = get_model_explainer()
predictor = get_default_predictor()

# Sample High-Risk Borrower
sample_borrower = {
    "customer_id": "C99999",
    "monthly_income": 45000,
    "requested_amount": 900000,
    "credit_score": 590,
    "credit_utilization": 0.78,
    "recent_delinquency": 1,
    "delinquency_count": 3,
    "active_loans": 3,
    "loan_type": "Personal_Loan"
}

pred = predictor.predict_single(sample_borrower)
local_shap = explainer.explain_local(sample_borrower)

print(f"Borrower Predicted Default Probability: {pred['default_probability']:.2%}")
print(f"Risk Category: {pred['risk_category']}")
print("\nTop Risk-Increasing SHAP Drivers:")
for item in local_shap['risk_increasing_factors']:
    print(f" + {item['feature']:25s} : +{item['shap_value']:.4f}")

print("\nTop Risk-Decreasing SHAP Drivers:")
for item in local_shap['risk_decreasing_factors']:
    print(f" - {item['feature']:25s} : {item['shap_value']:.4f}")
```
