# Notebook 04: Feature Engineering & Data Leakage Prevention

## Business Context & Methodological Rigor
Feature engineering transforms raw borrower records into predictive credit metrics. **Data leakage** occurs when information that is unavailable at decision time (e.g. post-origination repayment behavior or future DPD records) is improperly included in model training features. In this notebook, we construct strictly application-time financial features and execute chronological temporal splitting.

```python
import sys
import os
import pandas as pd
import numpy as np

sys.path.append(os.path.abspath('..'))
from ml.features import compute_application_features, apply_temporal_split

customers_df = pd.read_csv('../data/raw/customers.csv')
credit_df = pd.read_csv('../data/raw/credit_history.csv')
emp_df = pd.read_csv('../data/raw/employment_history.csv')
apps_df = pd.read_csv('../data/raw/applications.csv')

df = apps_df.merge(customers_df, on='customer_id') \
            .merge(credit_df, on='customer_id') \
            .merge(emp_df, on='customer_id')

# Compute application-time features
df_feat = compute_application_features(df)

print("Engineered Feature Summary:")
    print(df_feat[['dti', 'lti', 'emi_to_income', 'loan_burden_index', 'credit_score_norm']].describe())

train_df, val_df, test_df = apply_temporal_split(df_feat)
```
